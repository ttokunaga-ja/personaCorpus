# Personal GMV sanity check
Synthetic Mosaic Retail Analytics S2 • MTR-09 • completed order-line merchandise value, tax excluded, net recognized refunds.


```python
from decimal import Decimal
gross = Decimal('86.0')
refunds = Decimal('1.8')
net = gross - refunds
plan = Decimal('80.0')
prior = Decimal('78.6')
daily_net = [Decimal(x) for x in ('10.8','11.2','11.5','12.1','12.4','13.0','13.2')]
def run_once():
    return {'gross_m': str(gross), 'refunds_m': str(refunds), 'net_m': str(net), 'plan_variance_m': str(net-plan), 'prior_variance_m': str(net-prior), 'daily_total_m': str(sum(daily_net)), 'online_m': '46.3', 'stores_m': '37.9'}
run_1 = run_once()
run_2 = run_once()
assert run_1 == run_2
assert run_1['net_m'] == '84.2' and run_1['daily_total_m'] == '84.2'
run_1
```

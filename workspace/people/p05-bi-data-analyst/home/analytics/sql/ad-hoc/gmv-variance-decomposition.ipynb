# GMV variance decomposition
Synthetic Mosaic Retail Analytics | FY26 Q3 | 2026-07-13 | metric MTR-09

The controlled total is GMV ¥84.2M (84,200,000 JPY).

```python
gross_gmv = 86_000_000
refunds = 1_800_000
controlled_gmv = gross_gmv - refunds
plan_gmv = 80_000_000
prior_gmv = 78_600_000
print({'actual_yen': controlled_gmv})
print({'vs_plan_yen': controlled_gmv - plan_gmv})
```

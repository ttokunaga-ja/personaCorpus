# Channel mix exploration
Aggregate-only synthetic channel analysis for Mosaic Retail.

```python
channels = {'Online': 46_300_000}
channels['Stores'] = 37_900_000
total = sum(channels.values())
shares = {}
for name, value in channels.items():
    shares[name] = round(value / total, 3)
print(total)
print(shares)
```

```python
assert total == 84200000
assert sum(channels.values()) == total
max(channels, key=channels.get)
```

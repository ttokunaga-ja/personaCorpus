# Late-arrival sensitivity
A deterministic cutoff sensitivity for FY26 Q3 MTR-09 review.

```python
gross_gmv = 86_000_000
refunds_posted = 1_800_000
net_at_cut = gross_gmv - refunds_posted
print({'gross_gmv_yen': gross_gmv})
print({'net_at_cut_yen': net_at_cut})
```

```python
assert net_at_cut == 84_200_000
round(refunds_posted / gross_gmv * 100, 2)
```

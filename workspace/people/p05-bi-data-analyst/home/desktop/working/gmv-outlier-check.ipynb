# MTR-09 GMV outlier check

Synthetic S2 working analysis. MTR-09 is completed order-line merchandise value, tax excluded, net of recognized refunds, in JPY. Refresh through **2026-07-13 23:59 JST**; QA at **2026-07-14 06:00 JST**.


## Method

Daily values are inspected against a seven-day plan run-rate. The notebook is deterministic and contains no external I/O.


```python
daily = [10.8, 11.2, 11.5, 12.1, 12.4, 13.0, 13.2]
gross = 86.0
refunds = 1.8
plan = 80.0
gmv = sum(daily)
assert round(gmv, 1) == 84.2
assert round(gross - refunds, 1) == gmv
print(f"MTR-09 GMV: {gmv:.1f}M JPY")
print(f"Plan variance: {gmv-plan:.1f}M JPY ({(gmv-plan)/plan:.1%})")
```

```python
daily = [10.8, 11.2, 11.5, 12.1, 12.4, 13.0, 13.2]
run_rate = 80.0 / 7
for day, value in enumerate(daily, start=7):
    flag = "WATCH" if abs(value - run_rate) >= 1.5 else "OK"
    print(f"2026-07-{day:02d}: {value:.1f}M | variance {value-run_rate:+.1f}M | {flag}")
```

## Review conclusion

The late-week lift warrants a timing/completion check, but the daily total reconciles exactly to MTR-09 GMV (84.2M JPY). This is a review queue, not a defect finding.

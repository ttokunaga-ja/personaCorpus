## Method

The check reconciles gross merchandise value less recognized refunds to MTR-09. It deliberately distinguishes refund recognition from a cash-refund event date.


```python
gross = 86.0
recognized_refunds = 1.8
gmv = gross - recognized_refunds
channels = {"Online": (47.3, 1.0), "Stores": (38.7, 0.8)}
assert round(sum(g - r for g, r in channels.values()), 1) == gmv
print(f"Gross: {gross:.1f}M JPY")
print(f"Recognized refunds: {recognized_refunds:.1f}M JPY")
print(f"Net MTR-09 GMV: {gmv:.1f}M JPY")
```

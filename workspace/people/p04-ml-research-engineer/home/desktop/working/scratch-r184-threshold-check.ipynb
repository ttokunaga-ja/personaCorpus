# R184 threshold check

Offline calibration note for the 2,400-example evaluation set (S1: 1,600; S2: 800).

```python
scores = [0.91, 0.83, 0.74, 0.68, 0.59, 0.77, 0.71, 0.88]
threshold = 0.72
accepted = sum(score >= threshold for score in scores)
print(f'threshold={threshold:.2f} accepted={accepted / len(scores):.1%} review={1 - accepted / len(scores):.1%}')
```

## Working decision

Hold 0.72 for a manual S2 review; it leaves borderline scores in the review queue without accessing live data.
